```python
while True:
    try:
        data=input().split()
        a=int(data[0])
        b=int(data[1])
        maxed=max(a,b)
        mined=min(a,b)
        anser=1
        for i in range(mined,maxed+1):
            temp=1
            curr=i
            #需新變數不可直接用i進去加總外迴圈會亂掉
            while curr>1:
                if curr==1:
                    break
                if curr%2!=0:
                    curr=(curr*3)+1
                    temp+=1
                elif curr%2==0: #要用elif不然會重複判斷
                    curr=curr//2
                    temp+=1
            if temp>anser:
                anser=temp
        print(a,b,anser)
    except EOFError:
        break
```
