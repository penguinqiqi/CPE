```python
while True:
    try:
        carry=0
        temp=0
        data=input().split()
        a=int(data[0])
        b=int(data[1])
        if a==0 and b==0:
            break
        else:
            if len(str(a))>len(str(b)):
                n=len(str(a))
            else:
                n=len(str(b))
            for i in range(n):
                if a%10+b%10+temp>=10:
                    carry+=1
                    temp=1
                else:
                    temp=0
                a=a//10
                b=b//10
        if carry==0:
            print("No carry operation.")
        elif carry>1:
            print(carry,"carry operations.")
        else:
            print(carry,"carry operation.")
    except EOFError:
        break
```
